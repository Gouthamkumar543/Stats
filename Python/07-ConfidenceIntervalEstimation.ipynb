```python
import numpy as np
import pandas as pd
from scipy.stats import norm, t, chi2
data = pd.read_csv(r"C:\Users\gouth\Desktop\Goutham\Stats\DataSet\bank-data.csv")
data
```

```python
#Confidence Interval for Mean (σ known)
income = data["income"].values
n = len(income)
sigma = 10000  
confidence = 0.95
x_bar = np.mean(income)
z = norm.ppf((1 + confidence) / 2)
margin = z * (sigma / np.sqrt(n))
lower = x_bar - margin
upper = x_bar + margin
print("Mean (x̄):", x_bar)
print("95% CI for Mean (σ known):", (lower, upper))
```

```python
#Confidence Interval for Mean (σ unknown)
x_bar = np.mean(income)
s = np.std(income, ddof=1)
n = len(income)
confidence = 0.95
t_value = t.ppf((1 + confidence) / 2, df=n-1)
margin = t_value * (s / np.sqrt(n))
lower = x_bar - margin
upper = x_bar + margin
print("Mean (x̄):", x_bar)
print("Sample SD (s):", s)
print("95% CI for Mean (σ unknown):", (lower, upper))
```

```python
#Confidence Interval for Proportion
car = (data["car"].values == "YES").astype(int)
p = np.mean(car)
n = len(car)
confidence = 0.95
z = norm.ppf((1 + confidence) / 2)
margin = z * np.sqrt((p * (1 - p)) / n)
lower = p - margin
upper = p + margin
print("Sample Proportion (p):", p)
print("95% CI for Proportion:", (lower, upper))
```

```python
#Confidence Interval for Variance and Standard Deviation
confidence = 0.95
n = len(income)
s2 = np.var(income, ddof=1)
chi2_lower = chi2.ppf((1 - confidence) / 2, df=n-1)
chi2_upper = chi2.ppf((1 + confidence) / 2, df=n-1)
var_lower = (n - 1) * s2 / chi2_upper
var_upper = (n - 1) * s2 / chi2_lower
std_lower = np.sqrt(var_lower)
std_upper = np.sqrt(var_upper)
print("Sample Variance (s²):", s2)
print("95% CI for Variance:", (var_lower, var_upper))
print("95% CI for Standard Deviation:", (std_lower, std_upper))
```
