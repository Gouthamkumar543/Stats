```python
import numpy as np
import pandas as pd

np.random.seed(42)
n = 1000

#Uniform Distribution (0-10)
uniform_data = pd.DataFrame({
    "Satisfaction_Score": np.random.uniform(0, 10, n)
})

#Poisson Distribution (λ=5)
poisson_data = pd.DataFrame({
    "Daily_Applications": np.random.poisson(5, n)
})

#Normal Distribution (µ=200, σ=30)
normal_data = pd.DataFrame({
    "Cholesterol_Level": np.random.normal(200, 30, n)
})

#Exponential Distribution (scale=10)
exponential_data = pd.DataFrame({
    "Fraud_Wait_Time": np.random.exponential(10, n)
})

#Cauchy Distribution (clipped to [-10,10])
cauchy_data = pd.DataFrame({
    "Wine_Taste_Deviation": np.clip(np.random.standard_cauchy(n), -10, 10)
})

# Preview each dataset (first 5 rows)
print("Uniform Distribution:\n", uniform_data.head(), "\n")
print("Poisson Distribution:\n", poisson_data.head(), "\n")
print("Normal Distribution:\n", normal_data.head(), "\n")
print("Exponential Distribution:\n", exponential_data.head(), "\n")
print("Cauchy Distribution:\n", cauchy_data.head(), "\n")
```
