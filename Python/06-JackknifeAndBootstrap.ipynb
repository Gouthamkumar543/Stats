```python
import numpy as np
import pandas as pd
data = pd.read_excel(r"C:\Users\gouth\Desktop\Goutham\Stats\DataSet\Iris dataset.xlsx")
data
```

```python
#Jackknife Samples
sepal_length = data["sepal length"].values
n = len(sepal_length)
jackknife_means = []
for i in range(n):
    sample = np.delete(sepal_length, i)
    jackknife_means.append(np.mean(sample))
jackknife_means = np.array(jackknife_means)
original_mean = np.mean(sepal_length)
jackknife_mean = np.mean(jackknife_means)
bias = (n - 1) * (jackknife_mean - original_mean)
std_error = np.sqrt((n - 1) * np.mean((jackknife_means - jackknife_mean) ** 2))
print("Original Mean:", original_mean)
print("Jackknife Mean:", jackknife_mean)
print("Bias:", bias)
print("Standard Error:", std_error)
```

```python
#Bootstrap Samples
sepal_length = data["sepal length"].values
n = len(sepal_length)
B = 1000
bootstrap_means = []
for _ in range(B):
    sample = np.random.choice(sepal_length, size=n, replace=True)
    bootstrap_means.append(np.mean(sample))
bootstrap_means = np.array(bootstrap_means)
bootstrap_mean = np.mean(bootstrap_means)
std_error = np.std(bootstrap_means, ddof=1)  # unbiased estimate
lower = np.percentile(bootstrap_means, 2.5)
upper = np.percentile(bootstrap_means, 97.5)
print("Bootstrap Mean:", bootstrap_mean)
print("Standard Error:", std_error)
print("95% Confidence Interval:", (lower, upper))
```
